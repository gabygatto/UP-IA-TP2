```python
1. Objetivo

En esta etapa se cargan los datos sin etiquetar proporcionados para la entrega final. 
Se aplican las mismas transformaciones utilizadas durante el entrenamiento, se carga el modelo Random Forest seleccionado 
y se generan las predicciones de la variable objetivo. 
Finalmente se exporta un archivo con la nueva columna smoking_prediction.
```

2.Carga de librerias

```python
import pandas as pd
import joblib
```

3. Carga del DS sin etiquetas

```python
df_pred = pd.read_csv(
    "../data/raw/smoking_prediction_entrega.csv"
)

df_pred.head()
```

4. Verificacion de Estructura

```python
df_pred.info()
```

5.Copia de Trabajo

```python
df_final = df_pred.copy()
```

6. Aplicar exactamente el mismo preprocesamiento que al DS de entrenamiento

```python
#Eliminar columnas ID y oral 

df_final = df_final.drop(
    columns=["ID", "oral"]
)

#Codificar variables categoricas

df_final["gender"] = df_final["gender"].map({
    "F": 0,
    "M": 1
})

df_final["tartar"] = df_final["tartar"].map({
    "N": 0,
    "Y": 1
})
```

7. Verificacion de estructura final

```python
df_final.info()
```

8. Carga del modelo entrenado

```python
modelo = joblib.load(
    "../models/random_forest_final.joblib"
)

type(modelo)
```

9. Predicciones

```python
predicciones = modelo.predict(df_final)
```

10. Agregado de la columna Smoking_prediction

```python
df_pred["smoking_prediction"] = predicciones
```

```python
#Verificacion del agregado de columna predictiva
df_pred.head()
```

11. Validación de distribución de predicciones

```python
df_pred["smoking_prediction"].value_counts(normalize=True)*100
```

12. Export archivo final

```python
df_pred.to_csv(
    "../data/processed/predicciones_finales.csv",
    index=False
)
```
