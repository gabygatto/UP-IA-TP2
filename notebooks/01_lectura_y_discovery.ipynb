```python
for col in ["gender", "oral", "tartar"]:
    print(df[col].value_counts())
    print()

#Se identificó la variable oral como una característica constante, ya que presenta un único valor (Y) para los 50.000 
#registros. Debido a que no aporta capacidad discriminatoria para el modelo, 
#se considerará su eliminación durante la etapa de preprocesamiento.
    
```

```python
#Analisis de la columna Smoking

print(df["smoking"].value_counts())
df["smoking"].value_counts(normalize=True) * 100
```

```python
import matplotlib.pyplot as plt

df["smoking"].value_counts().plot(kind="bar")

plt.title("Distribución de la variable objetivo")
plt.xlabel("Smoking")
plt.ylabel("Cantidad")
plt.show()
```

La variable objetivo presenta un desbalance moderado. El 63,34% de los registros corresponde a personas no fumadoras y el 36,66% a personas fumadoras. Si bien existe una mayor representación de la clase 0, la diferencia no es extrema, por lo que inicialmente no se considera necesario aplicar técnicas de rebalanceo. Dado que el objetivo es identificar correctamente a los fumadores, la métrica principal será el F1-score de la clase positiva (smoking = 1).
